Instruction: Perform a single forward pass and compute for the error.

x = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}
y = \begin{bmatrix} 1 \end{bmatrix}
f=max(0, Z_n)

\text{hidden unit weights} = 
\begin{bmatrix}
w_{11} = 0.2 && w_{12} = -0.3 \\
w_{13} = 0.4 && w_{14} = 0.1 \\
w_{15} = -0.5 && w_{16} = 0.2
\end{bmatrix}

\text{output unit weights} = 
\begin{bmatrix}
w_{21} = -0.3 \\
w_{22} = -0.2
\end{bmatrix}

\theta = 
\begin{bmatrix}
\theta_{1} = -0.4 \\
\theta_{2} = 0.2 \\
\theta_{3} = 0.1
\end{bmatrix}

```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

# Inputs
x = np.array([1, 0, 1], dtype=float)   # shape (3,)
y = np.array([1], dtype=float)         # shape (1,)

# Hidden unit weights: 3 inputs -> 2 hidden units
W1 = np.array([
    [0.2, -0.3],
    [0.4,  0.1],
    [-0.5, 0.2]
])

# Output unit weights: 2 hidden units -> 1 output
W2 = np.array([
    [-0.3],
    [-0.2]
])

# Biases/thresholds:
# theta1, theta2 = hidden biases
# theta3 = output bias
theta = np.array([-0.4, 0.2, 0.1])

b1 = theta[:2]   # hidden biases
b2 = theta[2]    # output bias

# -------------------------
# Forward pass
# -------------------------

# Hidden layer
z1 = x @ W1 + b1
h = relu(z1)

# Output layer
z2 = h @ W2 + b2
y_hat = relu(z2)

# -------------------------
# Error
# -------------------------

error = (y - y_hat) ** 2
half_error = 0.5 * error

print("z1 =", z1)
print("h =", h)
print("z2 =", z2)
print("y_hat =", y_hat)
print("error =", error)
print("half_error =", half_error)
```

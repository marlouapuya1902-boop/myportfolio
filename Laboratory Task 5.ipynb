## Laboratory Task 5

### 1. Standard Imports

```python
import numpy as np
import torch
import random
```

### 2. Set Seed Function

```python
def set_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
```

### 3. NumPy Array

```python
set_seed(42)

arr = np.random.randint(0, 5, size=6)

print("arr:", arr)
print("dtype:", arr.dtype)
```

### 4. Tensor x From Array

```python
x = torch.from_numpy(arr)

print("x:", x)
print("dtype:", x.dtype)
print("shape:", x.shape)
```

### 5. Change int32 to int64

```python
x = x.to(torch.int64)

print("x:", x)
print("dtype:", x.dtype)
```

### 6. Reshape

```python
# Option A — .reshape()

x = x.reshape(3, 2)

print("x:\n", x)
print("shape:", x.shape)
```

### 7. Right Hand Column of x

```python
right_col = x[:, -1]

print("right column:", right_col)
```

### 8. Square Values

```python
# Option A — torch.square()

x_squared = torch.square(x)

print("x squared:\n", x_squared)
print("x unchanged:\n", x)
```

### 9. Create Tensor y Matrix-Multipliable With x

```python
set_seed(42)

y = torch.randint(0, 5, (2, 3))

print("y:\n", y)
print("shape:", y.shape)
```

### 10. Matrix Product

```python
result = x @ y

print("x @ y:\n", result)
print("shape:", result.shape)
```
